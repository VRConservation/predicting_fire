```python
# Create swipe map using leafmap (requires exactly 2 TIFF files)
if len(tif_files) < 2:
    print("⚠️  Need at least 2 TIFF files to create a swipe map.")
    print(f"Currently have: {len(tif_files)} file(s)")
else:
    # Use the first two files
    file1, file2 = tif_files[0], tif_files[1]
    
    with rasterio.open(file1) as src1:
        bounds1 = src1.bounds
        center_lat = (bounds1.top + bounds1.bottom) / 2
        center_lon = (bounds1.right + bounds1.left) / 2
    
    print(f"Creating swipe map with:")
    print(f"  Left layer: {Path(file1).name}")
    print(f"  Right layer: {Path(file2).name}")
    
    # Create split map
    m = leafmap.Map(center=[center_lat, center_lon], zoom=8)
    m.split_map(
        left_layer=file1,
        right_layer=file2,
        left_label=Path(file1).stem,
        right_label=Path(file2).stem
    )
    m.show()
```
