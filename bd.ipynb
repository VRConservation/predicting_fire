# Interactive Maps
Interactive maps to compare crown fire probability and potential impacts of fire on old-growth or mature forests and threatened and endangered species richness.

```python
# import dependencies
import os
import glob
from pathlib import Path
import rasterio
import numpy as np
import leafmap
```

```python
# Add raster datasets
crown = "tifs/zonal_clip.tif"
della = "tifs/della_clip.tif"
species = "tifs/spp_clip.tif"

# Initialize map
m = leafmap.Map(center=[39.12153746, -120.70667611], zoom=9)
m.add_basemap("CartoDB.Positron")
# Zoom to NV County and add layers to map
m.add_raster(della, colormap="viridis", layer_name="Old Growth", nodata=0)
m.add_raster(crown, colormap="rdylgn_r", layer_name="Crown Fire Probability", nodata=0)
m.set_center(lon=-120.70667611, lat=39.12153746, zoom=9)
```

```python
# Define palettes
crown_colors = ["#D7191C", "#FDAE61", "#FFFFBF", "#A6D96A", "#1A9641"]
crown_vmin = 0
crown_vmax = 1

della_colors = ["#440154", "#31688E", "#35B779", "#FDE724"]
della_vmin = 3
della_vmax = 12

# Add colorbar and map. Click wrench and layers icon to toggle layers
m.add_colorbar(colors=della_colors, vmin=della_vmin, vmax=della_vmax, caption="Old Growth")
m.add_colorbar(colors=crown_colors, vmin=crown_vmin, vmax=crown_vmax, caption="Crown Fire Probability")
m
```

```python
# Compare to listed species
# Create second map
m2 = leafmap.Map(center=[39.12153746, -120.70667611], zoom=9)
m2.add_basemap("CartoDB.Positron")
# Add layers to m2
m2.add_raster(species, colormap="reds", layer_name="Listed Species Richness", nodata=0)
m2.add_raster(crown, colormap="rdylgn_r", layer_name="Crown Fire Probability", nodata=0)
m2.set_center(lon=-120.70667611, lat=39.12153746, zoom=9)

# Define palettes for m2
crown_colors = ["#D7191C", "#FDAE61", "#FFFFBF", "#A6D96A", "#1A9641"]
crown_vmin = 0
crown_vmax = 1

# Continuous reds for species
species_colors = ["#FFF5F0", "#FEE0D2", "#FCBBA1", "#FC9272", "#FB6A4A", "#EF3B2C", "#CB181D", "#A50F15", "#67000D"]
species_vmin = 0
species_vmax = 9

# Add colorbars to m2
m2.add_colorbar(colors=species_colors, vmin=species_vmin, vmax=species_vmax, caption="Listed Species Richness")
m2.add_colorbar(colors=crown_colors, vmin=crown_vmin, vmax=crown_vmax, caption="Crown Fire Probability")
m2
```
