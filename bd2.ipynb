```python
# Split map using standard leafmap (ipyleaflet backend)
import leafmap

m = leafmap.Map(center=[39.12153746, -120.70667611], zoom=10, basemap="CartoDB.Positron")

della = "tifs/della_clip.tif"
crown = "tifs/zonal_clip.tif"

# Add split map comparison
m.split_map(
    left_layer=della,
    right_layer=crown,
    left_label="Old Growth",
    right_label="Crown Fire Probability",
    left_args={"colormap": "viridis", "nodata": 0},
    right_args={"colormap": "rdylgn_r", "nodata": 0},
)
m
```
